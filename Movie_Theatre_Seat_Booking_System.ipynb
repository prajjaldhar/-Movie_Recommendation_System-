Seat Booking Algorithm :

```python
def seatbooking(row,col):
  if row < 1 or row > 5 or col <1 or col >8:
    print("Invalid Seat Number!")
    return
  if seats[row-1, col-1] == 1:
    print("Seat Already Booked")
    return
  else:
    seats[row-1,col-1]=1
    print("Seat Booked Successfully!")
```

```python
seatbooking(2,3)
```

```python
seatbooking(5,2)
```

```python
seatbooking(-1,-3)
```
